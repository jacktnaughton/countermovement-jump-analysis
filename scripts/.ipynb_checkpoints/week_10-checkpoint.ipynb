```python
import sys
import os
sys.path.append("../")
print(os.getcwd())
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.manifold import TSNE
import seaborn as sns
import numpy as np
import pandas as pd
import torch
import os
from datetime import datetime
from src.model import VAE  

latent_dim = 4
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

vae_model_path = "../saved_models/vae20250408_230217_vae.pth"
vae = VAE(latent_dim=latent_dim).to(device)  

checkpoint = torch.load(vae_model_path, map_location=device)
vae.load_state_dict(checkpoint['model_state_dict'])
vae.eval()   
pd.set_option("display.max_columns", None)
```

```python
binned_males = pd.read_pickle("binned_males.pkl")  
```

```python
# 1. take average latent space of all males - DONE
# 2. reconstruct jump and calculate jump height from reconstruction and average of jump_duration_s
# 3. take top and bottom 10% per sport by this predicted jump height
# 4. 
```

```python
binned_males['vae_latent_space_4d'] = binned_males['vae_latent_space_4d'].apply(np.array)
latent_df = pd.DataFrame(binned_males['vae_latent_space_4d'].tolist(), index=binned_males.index)
latent_df.columns = ['z1', 'z2', 'z3', 'z4']
df_with_latents = pd.concat([binned_males[['profileID']], latent_df], axis=1)
mean_latent = df_with_latents.groupby('profileID')[['z1', 'z2', 'z3', 'z4']].mean()
mean_latent['avg_latent_space_4d'] = mean_latent.apply(lambda row: np.array([row['z1'], row['z2'], row['z3'], row['z4']]), axis=1)
binned_males['avg_latent_space_4d'] = binned_males['profileID'].map(mean_latent['avg_latent_space_4d'])
```

```python
def predict_jump_height_from_avg_latent(avg_latent_vec, jump_mean, jump_std, initial_val, vae):
    if initial_val < 1e-3:
        return np.nan  # skip bad values

    jump_t = jump_mean

    with torch.no_grad():
        z_tensor = torch.tensor(avg_latent_vec, dtype=torch.float32).unsqueeze(0)
        rec = vae.decode(z_tensor).squeeze().cpu().numpy()

    rec = np.clip(rec, 0, 5)
    offset = rec[0] * (20 / initial_val)
    adjusted = rec - offset
    net_force = adjusted - 1.0
    avg_net_force = np.mean(net_force)
    takeoff_velocity = avg_net_force * jump_t * 9.81
    height_inches = (takeoff_velocity ** 2) / (2 * 9.81) * 39.37

    return height_inches

# Step 1: Precompute group-wise aggregates
jump_std = binned_males["jump_duration_s"].std()
initial_values = binned_males.groupby("profileID")["initial_value"].mean()
avg_jump_duration = binned_males.groupby("profileID")["jump_duration_s"].mean()
avg_downsampled = binned_males.groupby("profileID")["downsampled_time_series"].apply(
    lambda x: np.mean(np.stack(x), axis=0)
)
team_by_profile = binned_males.groupby("profileID")["Team"].agg(
    lambda x: x.mode().iloc[0] if not x.mode().empty else np.nan
)

# Step 2: Compute predicted heights
mean_latent["average_predicted_jump_height"] = mean_latent.apply(
    lambda row: predict_jump_height_from_avg_latent(
        row["avg_latent_space_4d"],
        jump_mean,
        jump_std,
        initial_values.get(row.name, 1.0),
        vae
    ),
    axis=1
)

# Step 3: Merge all the group-wise values into mean_latent
mean_latent["avg_jump_duration_s"] = avg_jump_duration
mean_latent["avg_latent_space_4d"] = mean_latent[["z1", "z2", "z3", "z4"]].values.tolist()
mean_latent["Team"] = team_by_profile
mean_latent["avg_downsampled_time_series"] = avg_downsampled

# Step 4: Final cleaned DataFrame
final_df = mean_latent[[
    "avg_latent_space_4d",
    "avg_jump_duration_s",
    "average_predicted_jump_height",
    "avg_downsampled_time_series",
    "Team"
]].reset_index()
```

```python
def get_top_bottom_10_percent(df):
    results = []

    for team, group in df.groupby("Team"):
        if len(group) < 10:
            continue  

        lower_thresh = group["average_predicted_jump_height"].quantile(0.10)
        upper_thresh = group["average_predicted_jump_height"].quantile(0.90)

        bottom_10 = group[group["average_predicted_jump_height"] <= lower_thresh].copy()
        top_10 = group[group["average_predicted_jump_height"] >= upper_thresh].copy()

        bottom_10["performance_group"] = "Bottom 10%"
        top_10["performance_group"] = "Top 10%"

        results.append(bottom_10)
        results.append(top_10)

    return pd.concat(results)

top_bottom_df = get_top_bottom_10_percent(final_df)
```

```python
teams = top_bottom_df['Team'].unique()

for team in teams:
    df_team = top_bottom_df[top_bottom_df['Team'] == team]

    top_avg_latent = np.mean(np.stack(df_team[df_team['performance_group'] == 'Top 10%']['avg_latent_space_4d'].values), axis=0)
    bottom_avg_latent = np.mean(np.stack(df_team[df_team['performance_group'] == 'Bottom 10%']['avg_latent_space_4d'].values), axis=0)

    top_avg_height = df_team[df_team['performance_group'] == 'Top 10%']['average_predicted_jump_height'].mean()
    bottom_avg_height = df_team[df_team['performance_group'] == 'Bottom 10%']['average_predicted_jump_height'].mean()

    with torch.no_grad():
        top_latent_tensor = torch.tensor(top_avg_latent, dtype=torch.float32).unsqueeze(0)
        bottom_latent_tensor = torch.tensor(bottom_avg_latent, dtype=torch.float32).unsqueeze(0)

        top_reconstructed = vae.decode(top_latent_tensor).squeeze().cpu().numpy()
        bottom_reconstructed = vae.decode(bottom_latent_tensor).squeeze().cpu().numpy()

    top_reconstructed = np.clip(top_reconstructed, 0, 5)
    bottom_reconstructed = np.clip(bottom_reconstructed, 0, 5)

    top_jump_duration = df_team[df_team['performance_group'] == 'Top 10%']['avg_jump_duration_s'].mean()
    bottom_jump_duration = df_team[df_team['performance_group'] == 'Bottom 10%']['avg_jump_duration_s'].mean()

    plt.figure(figsize=(10, 6))
    plt.plot(top_reconstructed, label=f'Top 10% | Duration: {top_jump_duration:.2f}s | Pred Height: {top_avg_height:.2f} in', color='green')
    plt.plot(bottom_reconstructed, label=f'Bottom 10% | Duration: {bottom_jump_duration:.2f}s | Pred Height: {bottom_avg_height:.2f} in', color='red')

    plt.title(f'Reconstructed Jump Curve - {team}')
    plt.xlabel('Normalized Time')
    plt.ylabel('Force')
    plt.legend()
    plt.grid(True)
    plt.tight_layout()
    plt.show()
```

```python
teams = top_bottom_df['Team'].unique()
colors = plt.cm.get_cmap('tab10', len(teams))

max_duration_s = 0
all_forces = []

for team in teams:
    df_team = top_bottom_df[top_bottom_df['Team'] == team]
    
    top_duration = df_team[df_team['performance_group'] == 'Top 10%']['avg_jump_duration_s'].mean()
    bottom_duration = df_team[df_team['performance_group'] == 'Bottom 10%']['avg_jump_duration_s'].mean()
    max_duration_s = max(max_duration_s, top_duration, bottom_duration)
    
    with torch.no_grad():
        if not df_team[df_team['performance_group'] == 'Top 10%']['avg_latent_space_4d'].empty:
            top_avg_latent = np.mean(np.stack(df_team[df_team['performance_group'] == 'Top 10%']['avg_latent_space_4d'].values), axis=0)
            top_latent_tensor = torch.tensor(top_avg_latent, dtype=torch.float32).unsqueeze(0)
            top_reconstructed = vae.decode(top_latent_tensor).squeeze().cpu().numpy()
            all_forces.append(top_reconstructed)
        
        if not df_team[df_team['performance_group'] == 'Bottom 10%']['avg_latent_space_4d'].empty:
            bottom_avg_latent = np.mean(np.stack(df_team[df_team['performance_group'] == 'Bottom 10%']['avg_latent_space_4d'].values), axis=0)
            bottom_latent_tensor = torch.tensor(bottom_avg_latent, dtype=torch.float32).unsqueeze(0)
            bottom_reconstructed = vae.decode(bottom_latent_tensor).squeeze().cpu().numpy()
            all_forces.append(bottom_reconstructed)

max_duration_ms = max_duration_s * 1000

all_forces_concat = np.concatenate(all_forces)
min_force = np.min(all_forces_concat)
max_force = np.max(all_forces_concat)

plt.figure(figsize=(12, 7))
for i, team in enumerate(teams):
    df_team = top_bottom_df[top_bottom_df['Team'] == team]

    top_avg_latent = np.mean(np.stack(df_team[df_team['performance_group'] == 'Top 10%']['avg_latent_space_4d'].values), axis=0)
    top_jump_duration = df_team[df_team['performance_group'] == 'Top 10%']['avg_jump_duration_s'].mean()
    
    with torch.no_grad():
        top_latent_tensor = torch.tensor(top_avg_latent, dtype=torch.float32).unsqueeze(0)
        top_reconstructed = vae.decode(top_latent_tensor).squeeze().cpu().numpy()
    top_reconstructed = np.clip(top_reconstructed, 0, 5)

    x_top = np.linspace(0, top_jump_duration * 1000, len(top_reconstructed))

    plt.plot(x_top, top_reconstructed, label=team, color=colors(i))

plt.title('All Top 10% Reconstructed Jump Curves by Team')
plt.xlabel('Time (ms)')
plt.ylabel('Force')
plt.xlim(0, max_duration_ms + 50)
plt.ylim(min_force, max_force + 0.1)
plt.legend(title='Team', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True)
plt.tight_layout()
plt.show()


plt.figure(figsize=(12, 7))
for i, team in enumerate(teams):
    df_team = top_bottom_df[top_bottom_df['Team'] == team]

    bottom_avg_latent = np.mean(np.stack(df_team[df_team['performance_group'] == 'Bottom 10%']['avg_latent_space_4d'].values), axis=0)
    bottom_jump_duration = df_team[df_team['performance_group'] == 'Bottom 10%']['avg_jump_duration_s'].mean()
    
    with torch.no_grad():
        bottom_latent_tensor = torch.tensor(bottom_avg_latent, dtype=torch.float32).unsqueeze(0)
        bottom_reconstructed = vae.decode(bottom_latent_tensor).squeeze().cpu().numpy()
    bottom_reconstructed = np.clip(bottom_reconstructed, 0, 5)

    x_bottom = np.linspace(0, bottom_jump_duration * 1000, len(bottom_reconstructed))

    plt.plot(x_bottom, bottom_reconstructed, label=team, color=colors(i))

plt.title('All Bottom 10% Reconstructed Jump Curves by Team')
plt.xlabel('Time (ms)')
plt.ylabel('Force')
plt.xlim(0, max_duration_ms + 50)
plt.ylim(min_force, max_force + 0.1)
plt.legend(title='Team', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(14, 8))

lines = []
labels = []
heights = []

for team in teams:
    df_team = top_bottom_df[top_bottom_df['Team'] == team]

    top_group = df_team[df_team['performance_group'] == 'Top 10%']
    bottom_group = df_team[df_team['performance_group'] == 'Bottom 10%']

    if top_group.empty or bottom_group.empty:
        continue

    top_avg_latent = np.mean(np.stack(top_group['avg_latent_space_4d'].values), axis=0)
    bottom_avg_latent = np.mean(np.stack(bottom_group['avg_latent_space_4d'].values), axis=0)

    top_avg_height = top_group['average_predicted_jump_height'].mean()
    bottom_avg_height = bottom_group['average_predicted_jump_height'].mean()

    top_jump_duration = top_group['avg_jump_duration_s'].mean()
    bottom_jump_duration = bottom_group['avg_jump_duration_s'].mean()

    with torch.no_grad():
        top_latent_tensor = torch.tensor(top_avg_latent, dtype=torch.float32).unsqueeze(0)
        bottom_latent_tensor = torch.tensor(bottom_avg_latent, dtype=torch.float32).unsqueeze(0)

        top_reconstructed = vae.decode(top_latent_tensor).squeeze().cpu().numpy()
        bottom_reconstructed = vae.decode(bottom_latent_tensor).squeeze().cpu().numpy()

    top_reconstructed = np.clip(top_reconstructed, 0, 5)
    bottom_reconstructed = np.clip(bottom_reconstructed, 0, 5)

    x_top = np.linspace(0, top_jump_duration * 1000, len(top_reconstructed))
    x_bottom = np.linspace(0, bottom_jump_duration * 1000, len(bottom_reconstructed))

    # Plot Top 10%
    line_top, = plt.plot(x_top, top_reconstructed, 
                        label=f'{team} Top 10% ({top_avg_height:.1f} in, {top_jump_duration:.2f} s)', 
                        linestyle='-', alpha=0.8)
    lines.append(line_top)
    labels.append(f'{team} Top 10% ({top_avg_height:.1f} in, {top_jump_duration:.2f} s)')
    heights.append(top_avg_height)

    # Plot Bottom 10%
    line_bottom, = plt.plot(x_bottom, bottom_reconstructed, 
                           label=f'{team} Bottom 10% ({bottom_avg_height:.1f} in, {bottom_jump_duration:.2f} s)', 
                           linestyle='--', alpha=0.8)
    lines.append(line_bottom)
    labels.append(f'{team} Bottom 10% ({bottom_avg_height:.1f} in, {bottom_jump_duration:.2f} s)')
    heights.append(bottom_avg_height)

# Sort by predicted height descending
sorted_items = sorted(zip(heights, lines, labels), key=lambda x: x[0], reverse=True)
sorted_lines = [item[1] for item in sorted_items]
sorted_labels = [item[2] for item in sorted_items]

plt.title("Reconstructed Jump Curves for Top & Bottom 10% by Team")
plt.xlabel("Time (ms)")
plt.ylabel("Force")
plt.grid(True)

plt.legend(sorted_lines, sorted_labels, loc='upper left', fontsize='small', bbox_to_anchor=(1, 1))
plt.tight_layout()
plt.show()
```
